# Import Library yang dibutuhkan

```python
import pandas as pd
from google_play_scraper import reviews, Sort
from datetime import datetime
import time
from tqdm import tqdm
```

# Proses Pengambilan Data Ulasan PLN Mobile di Google Play Store

```python
print("Mengambil ulasan PLN Mobile (Google Play Store)")

google_reviews_all = []
continuation_token = None
iteration = 0

# App ID untuk PLN Mobile di Google Play Store
app_id = 'com.icon.pln123'

# Rentang waktu
start_date = datetime(2025, 9, 26)
end_date = datetime(2025, 10, 26)

# Proses Scraping Data Ulasan
while True:
    result, continuation_token = reviews(
        app_id,
        lang='id',
        country='id',
        sort=Sort.NEWEST,
        count=200,
        continuation_token=continuation_token
    )

# Filter dan Penyimpanan Data Ulasan
    if not result:
        print("Tidak ada data baru, hentikan loop.")
        break

    iteration += 1
    print(f"Iterasi {iteration}, {len(result)} ulasan baru...")

    # Simpan hanya ulasan pengguna (tanpa memperhitungkan balasan developer)
    for r in result:
        review_date = r['at']
        if start_date <= review_date <= end_date:
            google_reviews_all.append(r)
        elif review_date < start_date:
            print(f"⏹ Ditemukan ulasan sebelum {start_date.date()}, hentikan scraping.")
            continuation_token = None
            break

    # Menampilkan jumlah ulasan sementara
    print(f"✅ Total sementara: {len(google_reviews_all)} ulasan dalam rentang waktu.")

    if not continuation_token:
        break

    time.sleep(1)

print(f"Total ulasan pengguna antara {start_date.date()}–{end_date.date()}: {len(google_reviews_all)}")
```

# Konversi Data ke DataFrame

```python
# Konversi ke DataFrame dan pilih kolom penting SAJA
df_google = pd.DataFrame(google_reviews_all)[['userName', 'score', 'content', 'at']]
df_google.columns = ['nama', 'rating', 'ulasan', 'tanggal']

# Penyaringan Ulang Berdasarkan Tanggal
df_google = df_google[
    (df_google['tanggal'] >= start_date) & (df_google['tanggal'] <= end_date)
]

print(f"✅ Jumlah akhir ulasan pengguna: {len(df_google)}")
```

```python
print(df_google.columns)
```

# Statistik Deskriptif

```python
# Memastikan DataFrame tidak kosong
if not df_google.empty:
    print("\n📈 Ringkasan Data:")
    print(f"Total ulasan: {len(df_google)}")

    df_google['tanggal'] = pd.to_datetime(df_google['tanggal'], errors='coerce')
    print("Rentang tanggal:", df_google['tanggal'].min().date(), "hingga", df_google['tanggal'].max().date())

    print("Rata-rata rating:", round(df_google['rating'].mean(), 2))

    print("\nJumlah ulasan per bintang:")
    print(df_google['rating'].value_counts().sort_index(ascending=False))
else:
    print("⚠️ Tidak ada data ulasan yang ditemukan dalam rentang waktu tersebut.")
```

# Menyimpan Hasil ke CSV

```python
output_file = "ulasan_plnmobile_googleplaystore.csv"
df_google.to_csv(output_file, index=False, encoding='utf-8-sig')

print(f"💾 Data berhasil disimpan ke file: {output_file}")
print(f"📊 Total baris disimpan: {len(df_google)}")
```
