# Import Library

```python
import requests
import pandas as pd
from datetime import datetime
```

# Konfigurasi Awal

```python
# App ID untuk PLN Mobile di App Store
app_id = "1299581030"
# kode negara (id = Indonesia)
country = "id"
# jumlah halaman yang ingin diambil
max_pages = 200

print("Mengambil ulasan aplikasi PLN Mobile dari Apple App Store...\n")
all_reviews = []
```

# Proses Pengambilan Data (*Scraping*)

```python
print(f"Mengambil ulasan PLN Mobile (Apple App Store, {country.upper()})...\n")

for page in range(1, max_pages + 1):
    url = f"https://itunes.apple.com/{country}/rss/customerreviews/page={page}/id={app_id}/sortby=mostrecent/json"
    response = requests.get(url)

# Pemeriksaan Respons dan Validasi Data
    if response.status_code != 200:
        print(f"Gagal mengambil halaman {page}")
        break

    data = response.json()

    if "feed" not in data or "entry" not in data["feed"]:
        print(f"Tidak ada data di halaman {page}. Hentikan loop.")
        break

# Ekstraksi dan Penyimpanan Ulasan
    entries = data["feed"]["entry"][1:]
    print(f"Halaman {page} berhasil diambil ({len(entries)} ulasan).")

    for e in entries:
        nama = e["author"]["name"]["label"]
        rating = e["im:rating"]["label"]
        ulasan = e["content"]["label"]
        tanggal = e["updated"]["label"]

        all_reviews.append({
            "nama": nama,
            "rating": int(rating),
            "ulasan": ulasan,
            "tanggal": tanggal
        })

# Ringkasan Jumlah Ulasan
print(f"\n📊 Total ulasan berhasil diambil: {len(all_reviews)}")
```

# Konversi Data ke DataFrame

```python
# Konversi list of dicts menjadi DataFrame
df = pd.DataFrame(all_reviews)

# Menampilkan 5 data pertama
print(df.head())

# Mengubah kolom 'tanggal' menjadi datetime
df['tanggal'] = pd.to_datetime(df['tanggal'])

# Mengambil tanggal awal dan akhir
tanggal_awal = df['tanggal'].min()
tanggal_akhir = df['tanggal'].max()

print(f"Rentang tanggal ulasan: {tanggal_awal.date()} s/d {tanggal_akhir.date()}")
```

# Analisis Deskriptif Rating

```python
# Menghitung jumlah ulasan per rating
jumlah_per_bintang = df['rating'].value_counts().sort_index()

print("Jumlah ulasan per bintang:")
print(jumlah_per_bintang)

print(f"Rating rata-rata: {df['rating'].mean():.2f}")

print(f"Total ulasan: {len(df)}")
```

# Menyimpan Data ke File CSV

```python
df = pd.DataFrame(all_reviews)
df['tanggal'] = pd.to_datetime(df['tanggal'])
df.to_csv("ulasan_plnmobile_applestore.csv", index=False, encoding="utf-8-sig")

print("💾 Data berhasil disimpan ke 'ulasan_plnmobile_applestore.csv'")
```

# Proses Filtering Ulasan Berbahasa Indonesia


```python
def deteksi_bahasa(teks):
    try:
        return detect(str(teks))
    except LangDetectException:
        return "unknown"

df["bahasa"] = df["ulasan"].apply(deteksi_bahasa)

print("\nDistribusi bahasa:")
print(df["bahasa"].value_counts())

# Mengambil ulasan berbahasa Indonesia
df_id = df[df["bahasa"] == "id"]

print(f"\n Jumlah ulasan berbahasa Indonesia: {len(df_id)}")
```

```python
# Mengambhil hanya ulasan berbahasa Indonesia
df_id = df[df["bahasa"] == "id"]

print(f"Jumlah ulasan berbahasa Indonesia: {len(df_id)}")

kolom_dipakai = ["nama", "rating", "ulasan", "tanggal"]
df_id = df_id[kolom_dipakai]

# Simpan ke CSV
output_file = "ulasan_plnmobile_applestore.csv"
df_id.to_csv(output_file, index=False, encoding="utf-8-sig")

from google.colab import files
files.download(output_file)
```
