```python
import pandas as pd
from google.colab import files

# =====================================================
# Upload file CSV
# =====================================================
uploaded = files.upload()
file_names = list(uploaded.keys())

print("Urutan file yang akan dibaca:")
for i, f in enumerate(file_names, 1):
    print(f"{i}. {f}")

df_list = []

# =====================================================
# Membaca dan menggabungkan setiap file CSV
# =====================================================
for file in file_names:
    try:
        df = pd.read_csv(
            file,
            encoding="utf-8",
            quotechar='"',
            doublequote=True,
            escapechar=None,
            on_bad_lines='warn'
        )
    except Exception as e:
        print(f"Error membaca {file}: {e}")
        continue

    # Menambahkan kolom sumber
    df["sumber"] = file

    # Konversi kolom tanggal ke datetime
    if 'tanggal' in df.columns:
        df['tanggal'] = pd.to_datetime(df['tanggal'], errors='coerce')

    # Mengurutkan berdasarkan tanggal
    if 'tanggal' in df.columns:
        df = df.sort_values('tanggal', ascending=True).reset_index(drop=True)

    print(f"{file}: {len(df)} baris berhasil dibaca")
    if 'tanggal' in df.columns:
        print(f"  Tanggal paling awal : {df['tanggal'].min()}")
        print(f"  Tanggal paling akhir: {df['tanggal'].max()}\n")

    df_list.append(df)

print("=" * 70)
print(f"Jumlah file yang dibaca: {len(df_list)}\n")

# =====================================================
# Menggabungkan seluruh DataFrame
# =====================================================
df_gabungan = pd.concat(df_list, ignore_index=True)

print(f"Total data setelah digabung: {len(df_gabungan)}")

# Standarisasi nama kolom
df_gabungan = df_gabungan.rename(columns={
    "userName": "nama",
    "score": "rating",
    "content": "ulasan",
    "at": "tanggal"
})

# =====================================================
# ANALISIS DESKRIPTIF DATASET
# =====================================================
print("\nANALISIS DESKRIPTIF DATASET")

# Jumlah baris dan kolom
print(f"Jumlah baris : {df_gabungan.shape[0]}")
print(f"Jumlah kolom : {df_gabungan.shape[1]}\n")

# Informasi tipe data
print("Informasi Tipe Data:")
df_gabungan.info()

# Ringkasan deskriptif per kolom
print("\nRingkasan Deskriptif per Kolom:")
deskriptif = pd.DataFrame({
    "Kolom": df_gabungan.columns,
    "Tipe Data": df_gabungan.dtypes.astype(str),
    "Count": df_gabungan.count(),
    "Unique": df_gabungan.nunique(),
    "Missing Values": df_gabungan.isnull().sum()
})

display(deskriptif)

# =====================================================
# Informasi tambahan
# =====================================================
print("\nDistribusi data berdasarkan sumber:")
print(df_gabungan['sumber'].value_counts())

print("\n5 data pertama:")
display(df_gabungan.head())

print("\n5 data terakhir:")
display(df_gabungan.tail())

# =====================================================
# Simpan hasil
# =====================================================
output_file = "ulasan_plnmobile.csv"
df_gabungan.to_csv(output_file, index=False, encoding="utf-8-sig")

print(f"\nFile disimpan sebagai: {output_file}")
files.download(output_file)
```
